```python
# Checking for outliers

def detect_outliers_z_score(df, threshold=3):
    # Calculate the Z-scores for each numeric column
    numeric_cols = df.select_dtypes(include=[np.number])  # Only numeric columns
    z_scores = np.abs((numeric_cols - numeric_cols.mean()) / numeric_cols.std())
    
    # Identify outliers (z-score > threshold)
    outliers = (z_scores > threshold)
    
    return outliers

# Check for outliers in the numeric columns of the dataset
outliers_df = detect_outliers_z_score(df_read)

# Print the rows with outliers for inspection
print("Outliers in the numeric columns of the dataset:\n", df_read[outliers_df.any(axis=1)])
```

```python
# Define the function to categorize the age groups
def categorize_age(age):
    if age >= 18 and age <= 34:
        return 'Young Adult'
    elif age >= 35 and age <= 64:
        return 'Adult'
    elif age >= 65:
        return 'Older Adult'
    else:
        return 'Unknown'  

# Create new variable for Age Group
df_read['Age_Group'] = df_read['Age'].apply(categorize_age)
```

#### Create lifestyle score or metabolic syndrome score.
**Metabolic Syndrome Score** <br>
Metabolic Syndrome is a group of conditions that raises a persons risk of heart disease. The Metabolic Syndrome Score applys all of these conditions to an easy to understand condition. To be diagnosed, an individual must have at least three of five *specific* risk factors: a large waistline (differs by sex), high triglycerides (*not avaliable in this dataset*), a low HDL cholesterol (differs by sex), high blood pressure, and high fasting blood sugar.  

The score is based on five key components commonly associated with metabolic syndrome:
Waist Circumference: Abdominal obesity is a key indicator of metabolic risk. Increased waist circumference, particularly in individuals with excess visceral fat, is linked to higher risks for cardiovascular disease and diabetes. <br>
The higher the score, the higher chances of metabolic syndrome. (Up to 4)

```python
# Create lifestyle score by combining Physical Activity Level, Daily Caloric Intake, Alcohol Consumption, and Smoking Status
def metabolic_syndrome_score(row): 
    score = 0

    # Adding Waist Circumference (defined in centimeters) point
    if row['Waist_Circumference'] >= (102 if row['Sex'] == 'Male' else 88):
        score += 1

    # Adding Blood Pressure Point
    if row['Blood_Pressure_Systolic'] >= 130 or row['Blood_Pressure_Diastolic']:
        score += 1

    # Fasting Blood Glucose (mg/dL) 
    if row['Fasting_Blood_Glucose'] >= 100:  # Threshold for elevated fasting blood glucose
        score += 1

    # Adding HDL Cholesterol (mg/dL)
    if (row['Sex'] == 'Male' and row['G_Cholesterol_HDL'] < 40) or (row['Sex'] == 'Female' and row['G_Cholesterol_HDL'] < 50):
        score += 1

    return score

# Apply the function to our dataset to create the metabolic syndrome score
df_cleaned['Metabolic_Syndrome_Score'] = df_cleaned.apply(metabolic_syndrome_score, axis=1)

# Display results
df_cleaned
```

**Lifestyle Syndrome Score** <br>
This hypothetical score will be put together using key factors using an individuals physical activity level, dietary intake, alcohol consunption level, as well as their smoking status. All of these factors directly impact a person's overall health. <br>
The higher the score, the healthier the lifestyle. (Up to 4)

```python
# Create metabolic syndrome score by combining Waist Circumference, Blood Pressure, Fasting Blood Glucose and HDL Cholesterol
def lifestyle_score(row):
    score = 0
    
    # Physical Activity Level
    if row['Physical_Activity_Level'] in ['Moderate', 'High']:
        score += 1

    # Dietary Intake Calories (fl) for Male and Female 
    if row['Sex'] == 'Female':
        if 1600 <= row['Daily_Calorie_Intake'] <= 2400: # Generalized for adults from NIH
            score += 1
    elif row['Sex'] == 'Male':
        if 2000 <= row['Daily_Calorie_Intake'] <= 3200: # Generalized for adults from NIH
            score += 1
    
    # Alcohol Consumption
    if row['Alcohol_Consumption'] == 'No':
        score += 1
    
    # Smoking Status
    if row['Smoking_Status'] == 'Non-smoker':
        score += 1
    
    return score

# Apply the function to create a new column 'Lifestyle_Score'
df_cleaned['Lifestyle_Score'] = df_cleaned.apply(lifestyle_score, axis=1)

#Show results
df_cleaned
```

Throughout the analysis process, it became abundantly clear that strong, direct correlations among individuals, even when specifying metabolic syndrome scores, was difficult to identify when viewing the dataset as a whole. * Waist Circumference: Individuals with higher metabolic syndrome scores may be at greater risk for central obesity, which is commonly linked to metabolic dysfunction. (Up to -0.41)

* Fasting Blood Glucose: Fasting blood glucose levels are highly variable across the sample, making it difficult to identify a distinct pattern without further analysis.

* Smoking and Metabolic Syndrome: Smoking alone may not be a significant factor in determining metabolic syndrome risk.

* Lifestyle Score and Metabolic Syndrome: Those with more severe metabolic syndrome may attempt to improve their lifestyle choices.
